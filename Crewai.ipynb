```python
import os
from crewai import LLM, Agent, Task, Crew
os.environ["GEMINI_API_KEY"] = "YOUR_GEMINI_API_KEY"
llm = LLM(
    model="gemini/gemini-3.5-flash",
    temperature=0.2,
    max_completion_tokens=1000,
    top_p=0.9,
    stream=False
)
```

```python
resume_parser = Agent(
    role="reume parser",
    goal="Extract skills, experiences, and achievements from resumes",
    backstory= "Trained in advanced NLP to handle messy, multi‑format resumes (PDF, DOCX, LinkedIn exports), Skilled at extracting structured data: skills, education, achievements, and work history, Designed to preserve context — even from complex layouts or non‑standard formatting, Converts unstructured text into clean JSON for downstream analysis, Acts as the “data foundation” for the entire crew, ensuring accuracy from the start.",
    llm=llm
)
```

```python
job_mathcer = Agent(
    role="job matcher",
    goal="Evaluate resumes against job descriptions to identify strengths, gaps, and ATS keyword alignment.",
    backstory= "Expert in ATS (Applicant Tracking System) keyword matching and recruiter workflows, Compares parsed resumes against job descriptions to highlight strengths and gaps, Calculates a match score to quantify candidate fit, Identifies transferable skills that may not be obvious to recruiters, Provides actionable insights on what to add or emphasize for better alignment.",
    llm=llm
)
```

```python
keyword_finder = Agent(
    role="keyword optimizer/finder",
    goal="Generate ATS-friendly bullet points and rephrase achievements using industry-standard keywords.",
    backstory="Specializes in rewriting achievements into impact‑driven bullet points, Embeds industry‑specific terminology naturally for ATS compliance, Focuses on measurable outcomes (e.g., “Improved efficiency by 30%”), Balances recruiter readability with machine keyword scanning, Ensures resumes stand out in both automated filters and human reviews.",
    llm=llm
)
```

```python
formating = Agent(
    role="reume formatter",
    goal="Produce polished, professional resumes tailored to different industries and roles.",
    backstory="Produce polished, professional resumes tailored to different industries and roles, backstory= Applies modern, recruiter‑friendly design standards, Adapts layouts for different industries (Data Science, IT, Business, Creative), Ensures readability with clean fonts, spacing, and section hierarchy, Outputs multiple polished versions (PDF, Word) tailored to specific roles, Acts as the “presentation layer” — making resumes visually professional and appealing.",
    llm=llm
)
```

```python
interview_prep  = Agent(
    role="interview prep coach",
    goal="Generate tailored interview questions and STAR-method answer prompts based on job descriptions.",
    backstory= "Generates tailored interview questions based on job descriptions, Uses the STAR method (Situation, Task, Action, Result) to guide structured answers, Covers both technical and behavioral scenarios, Helps candidates anticipate recruiter expectations and prepare confidently, Acts as a mentor, turning resume content into strong talking points for interviews.",
    llm=llm
)
```

```python
parse_resume_task = Task(
    description="Analyze the provided resume in depth and extract every relevant piece of information into a structured and organized profile. Understand the context of each section rather than simply copying text. Identify the candidate's contact information, professional summary, technical skills, soft skills, work experience, projects, education, certifications, achievements, awards, leadership experience, extracurricular activities, programming languages, frameworks, tools, and technologies. Preserve the original meaning while correcting obvious formatting inconsistencies and removing duplicate information. Do not assume or invent any information that is not explicitly available in the resume. Produce a clean, detailed, and machine-readable profile that serves as the foundation for the remaining agents.",
    expected_output="A comprehensive and well-structured candidate profile containing every important detail extracted from the resume, organized into logical sections with complete accuracy and consistency.",
    agent=resume_parser
)
```

```python
job_matching_task = Task(
    description="Carefully compare the structured resume against the provided job description and evaluate how well the candidate matches the role. Analyze technical skills, programming languages, frameworks, work experience, education, certifications, projects, domain knowledge, responsibilities, achievements, and ATS keywords. Identify the candidate's strongest qualifications, missing skills, transferable skills, and areas that require improvement. Estimate an ATS compatibility score based on industry best practices and explain the reasoning behind every recommendation. Prioritize improvements that will have the greatest impact on increasing the candidate's chances of passing ATS screening and impressing recruiters.",
    expected_output="A detailed ATS compatibility report containing an overall match score, strengths, weaknesses, missing keywords, missing technical skills, transferable skills, recruiter observations, and a prioritized list of recommendations for improving the resume.",
    agent=job_mathcer
)
```

```python
keyword_optimization_task = Task(
    description="Rewrite and optimize the resume using the findings from the ATS analysis while maintaining complete factual accuracy. Strengthen the professional summary, improve work experience descriptions, rewrite project descriptions into achievement-oriented bullet points, naturally incorporate relevant ATS keywords, improve clarity, eliminate redundancy, and increase recruiter appeal. Use powerful action verbs, measurable accomplishments whenever possible, and industry-standard terminology. Ensure every improvement reflects the candidate's actual experience without exaggerating or fabricating qualifications.",
    expected_output="A fully optimized ATS-friendly version of the resume with improved summaries, stronger experience descriptions, enhanced project sections, optimized skills, and naturally integrated industry keywords while preserving factual correctness.",
    agent=keyword_finder
)
```

```python
format_resume_task = Task(
    description="Transform the optimized resume into a polished, modern, and recruiter-friendly document. Organize every section in a logical order that maximizes readability and professionalism. Maintain consistent spacing, headings, typography, bullet formatting, and alignment while ensuring the layout remains ATS compatible. Avoid complex tables, graphics, columns, or formatting that could interfere with applicant tracking systems. Produce a resume that is visually appealing, easy to scan within a few seconds, and suitable for professional applications across the target industry.",
    expected_output="A professionally formatted resume with a clean ATS-compatible layout, consistent styling, clear section hierarchy, excellent readability, and presentation quality suitable for immediate submission.",
    agent=formating
)
```

```python
interview_preparation_task = Task(
    description="Create a personalized interview preparation guide using the optimized resume and the target job description. Analyze the candidate's experience and anticipate the questions that interviewers are most likely to ask. Generate technical questions, behavioral questions, project-based discussions, role-specific scenarios, and HR screening questions. Provide STAR-method guidance for behavioral answers, identify topics the candidate should revise before the interview, explain important technical concepts that may be discussed, suggest thoughtful questions the candidate can ask the interviewer, and provide practical interview strategies that improve confidence and performance throughout the hiring process.",
    expected_output="A comprehensive interview preparation guide containing technical interview questions, behavioral questions, STAR answer frameworks, project discussion topics, revision recommendations, interview strategies, and role-specific preparation advice tailored to the target position.",
    agent=interview_prep
)
```

```python
resume_strategist_crew = Crew(
    agents = [Resume_Parser, job_mathcer, keyword_finder, formating, interview_prep],
    tasks=[parse_resume_task, job_matching_task, keyword_optimization_task, format_resume_task, interview_preparation_task],
    verbose=True
)
```

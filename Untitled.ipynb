```python
%%sql
CREATE TABLE customer_transactions (
    transaction_id   INT PRIMARY KEY AUTO_INCREMENT, -- unique ID for each transaction
    customer_id      INT NOT NULL,                   -- links to customer table
    transaction_date DATETIME NOT NULL,              -- when the transaction occurred
    amount           DECIMAL(10,2) NOT NULL,         -- transaction amount
    payment_method   VARCHAR(50),                    -- e.g., Credit Card, UPI, Cash
    status           VARCHAR(20),                    -- e.g., Completed, Pending, Failed
    remarks          VARCHAR(255)                    -- optional notes
);
```

```python
%%sql
-- Create Restaurant table
CREATE TABLE restaurant (
    id INT PRIMARY KEY,
    name VARCHAR(100),
    location VARCHAR(100)
);

-- Create Orders table
CREATE TABLE orders (
    id INT PRIMARY KEY,
    restaurant_id INT,
    order_date DATE,
    FOREIGN KEY (restaurant_id) REFERENCES restaurant(id)
);

-- Insert sample data into Restaurant

(1, 'Spice Garden', 'Whitefield'),
(2, 'Ocean Breeze', 'MG Road'),
(3, 'Royal Tandoor', 'Koramangala');

-- Insert sample data into Orders
INSERT INTO orders (id, restaurant_id, order_date) VALUES
(101, 1, '2026-06-01'),
(102, 2, '2026-06-02'),
(103, 1, '2026-06-03'),
(104, 3, '2026-06-04');
```

```python
%%sql 
DROP TABLE IF EXISTS mark;
-- Create the table
CREATE TABLE mark (
    st_id INT PRIMARY KEY,
    st_name VARCHAR(50),
    marks INT
);

-- Insert sample records
INSERT INTO mark (st_id, st_name, marks) 
VALUES 
    (1, 'dharsan', 77),
    (2, 'drishya', 88),
    (3, 'christy', 76),
    (4, 'krithi', 77);

```

```python
%%sql

select * from mark
```

```python
%%sql

select st_id, st_name, marks,
rank() over (order by marks desc) as orderd_mark,
dense_rank() over (order by marks desc) as orderd_mark_dense,
percent_rank() over (order by marks desc) as orderd_mark_percent from mark
```
